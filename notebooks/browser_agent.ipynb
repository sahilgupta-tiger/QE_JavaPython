```python
from browser_use import Agent, Browser, BrowserProfile, ChatAzureOpenAI
import asyncio
```

```python
llm_model = ChatAzureOpenAI(api_version="2024-10-21",
                azure_endpoint="https://tigeropenaiservice.openai.azure.com/",
                model="gpt-4o",
                api_key="4feeebf652bd46168c6a863b99314fc9")

browser = Browser(
    browser_profile=BrowserProfile(
        headless=False,
        extra_chromium_args=[
            "--no-sandbox",
            "--disable-dev-shm-usage",
            "--disable-gpu",
        ]
    )
)
```

```python
task = """Login and then extract the data from all the charts in structured format.

1. Enter this username: your_email_here.
2. Click on Select and wait for user to complete okta.
3. Continue then and wait for the page to load.
4. Extract the data from the top 2 charts.

Return JSON with this structure:
[{"chart_heading": "value", "data": {"kpi_name": "kpi_value"}}]"""

agent = Agent(
    task=task,
    llm=llm_model,
    initial_actions=[{"navigate": {"url": url, "new_tab": True}}],
    browser=browser,
)
history = await agent.run()
```
